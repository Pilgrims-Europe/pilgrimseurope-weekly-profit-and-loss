```python
from pathlib import Path

import pandas as pd
import snowflake.connector

SQL_DIR = Path.cwd().parent / "sql"
MAX_ROWS = 200

# Uses the default connection in your Snowflake connections.toml. Read-only work only.
conn = snowflake.connector.connect(connection_name="default")


def run(sql: str) -> pd.DataFrame:
    """Run SQL and return at most MAX_ROWS rows."""
    cur = conn.cursor()
    try:
        cur.execute(sql)
        return cur.fetch_pandas_all().head(MAX_ROWS)
    finally:
        cur.close()


def run_file(name: str) -> pd.DataFrame:
    """Run sql/<name>.sql."""
    return run((SQL_DIR / f"{name}.sql").read_text(encoding="utf-8"))
```
